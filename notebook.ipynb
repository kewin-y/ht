```python
from ht.core import Tensor
import numpy as np
import torch
```

```python
x = Tensor([[1,2,3]], _label='x', requires_grad=True)
y = Tensor([1,2,3], _label='y', requires_grad=True)

z = x @ y
```

```python
z.backward()
```

```python
x.grad
```

```python
y.grad
```

```python
xx = torch.tensor([[1,2,3]], dtype=float)
yy = torch.tensor([1,2,3], dtype=float)
xx.requires_grad = True
yy.requires_grad = True
```

```python
zz = xx @ yy
```

```python
zz.backward()
```

```python
xx.grad
```

```python
yy.grad
```
